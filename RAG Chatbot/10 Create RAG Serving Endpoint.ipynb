## 1. Crear el endpoint de Serving desde el modelo registrado

```python
from mlflow.deployments import get_deploy_client

client = get_deploy_client("databricks")

try:
    endpoint = client.create_endpoint(
        name="appliance_chatbot",  
        config={
            "served_entities": [
                {
                    "name": "appliance_chatbot-entity",  
                    "entity_name": "bluetab.rag.appliance_chatbot_model",
                    "entity_version": "1",
                    "workload_size": "Small",
                    "scale_to_zero_enabled": True
                }
            ],
            "traffic_config": {
                "routes": [
                    {
                        "served_model_name": "appliance_chatbot-entity",
                        "traffic_percentage": 100
                    }
                ]
            }
        }
    )
except Exception as e:
    print(f"Error creating endpoint: {e}")
```
